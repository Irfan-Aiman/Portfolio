```python
import pandas as pd
from google.colab import drive

#connect google drive
drive.mount('/content/drive')

#load dataset
file_path = '/content/drive/MyDrive/Portfolio/Portfolio 1/Superstore.csv'
df = pd.read_csv(file_path, encoding='ISO-8859-1')

#display the first few rows of the dataset
print(df.head())
```

to list the column's names

```python
# Print the shape (Rows, Columns) to confirm it loaded correctly
print(f"Dataset shape: {df.shape}")

# Inspect actual column names
print("\nColumns in your dataset:")
print(df.columns.tolist())
```

### Check & Fix Data Types
Usually, date columns get imported as text (object). Convert them to proper datetime objects so you can perform monthly/yearly trend analysis later.

```python
# Convert date columns
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])

# Verify data types
df.dtypes
```

### Check for Missing Values & Duplicates
Verify if any rows are repeated or if key fields have blank values.

```python
#check for Null Values
print ("Missing Values:")
print(df.isnull().sum())

#check for Duplicates
print ("\nDuplicates:")
print(df.duplicated().sum())

# Drop duplicate rows if any exist
df = df.drop_duplicates()
```

### Engineer Essential Business Metrics
To answer the core business questions (like profit margin and discount impact), create derived columns:

```python
# 1. Profit Margin (%) = (Profit / Sales) * 100
df['Profit Margin %'] = (df['Profit'] / df['Sales']) * 100

# 2. Extract Year and Month for time-series dashboard charts
df['Order Year'] = df['Order Date'].dt.year
df['Order Month'] = df['Order Date'].dt.to_period('M')

# 3. Calculate Unit Price and Unit Cost
df['Unit Price'] = df['Sales'] / df['Quantity']
df['Unit Cost'] = (df['Sales'] - df['Profit']) / df['Quantity']
```

### Export Cleaned Data
Once cleaned, export the dataset to a clean CSV file. This cleaned file will be what you load into Power BI for your interactive dashboard or import into SQL for database queries.

```python
df.to_csv('/content/drive/MyDrive/Portfolio/Portfolio 1/Superstore_cleaned.csv', index=False)
print("Cleaned data exported to Superstore_cleaned.csv")
```
